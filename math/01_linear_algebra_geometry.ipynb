## Dot products and geometry

For vectors $\mathbf{x},\mathbf{w}\in\mathbb{R}^d$,

$$
\mathbf{x}^T\mathbf{w}=\|\mathbf{x}\|\|\mathbf{w}\|\cos\theta.
$$

So a linear model's score is both a weighted sum **and** a geometric projection.

```python
import numpy as np, matplotlib.pyplot as plt
x=np.array([2.,1.]); w=np.array([1.,2.])
dot=x@w
cos=dot/(np.linalg.norm(x)*np.linalg.norm(w))
print("dot:",dot,"cos(theta):",cos,"angle deg:",np.degrees(np.arccos(cos)))
```

## Orthogonal projection

Projection of $\mathbf{x}$ onto a nonzero vector $\mathbf{u}$:

$$
\operatorname{proj}_{\mathbf{u}}(\mathbf{x})
=
\frac{\mathbf{x}^T\mathbf{u}}{\mathbf{u}^T\mathbf{u}}\mathbf{u}.
$$

```python
u=np.array([1.,1.]); x=np.array([3.,1.])
proj=(x@u)/(u@u)*u
res=x-proj
print("projection:",proj,"residual:",res,"orthogonal check:",res@u)
```

## Rank and information

The rank of a matrix is the number of linearly independent directions it contains.

A low-rank matrix can be represented compactly:

$$
W\approx U_r\Sigma_rV_r^T.
$$

This idea appears in PCA, matrix factorization, compression, and LoRA.

```python
rng=np.random.default_rng(0)
A=rng.normal(size=(20,3))@rng.normal(size=(3,15))
print("shape:",A.shape,"rank:",np.linalg.matrix_rank(A))
U,S,Vt=np.linalg.svd(A,full_matrices=False)
A2=U[:,:3]@np.diag(S[:3])@Vt[:3]
print("rank-3 reconstruction MSE:",np.mean((A-A2)**2))
```

## Eigenvectors and invariant directions

If

$$
A\mathbf{v}=\lambda\mathbf{v},
$$

then applying $A$ to $\mathbf{v}$ changes only magnitude, not direction. PCA uses eigenvectors of a covariance matrix.

```python
A=np.array([[3.,1.],[1.,3.]])
vals,vecs=np.linalg.eigh(A)
for i in range(2):
    print(vals[i],np.allclose(A@vecs[:,i],vals[i]*vecs[:,i]))
```
