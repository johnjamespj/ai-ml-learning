## Derivative as local sensitivity

For scalar $f(x)$,

$$
f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}.
$$

For $f:\mathbb{R}^d\to\mathbb{R}$, the gradient is

$$
\nabla f=
\begin{bmatrix}
\partial f/\partial x_1\\
\vdots\\
\partial f/\partial x_d
\end{bmatrix}.
$$

The gradient points in the direction of steepest local increase.

```python
import numpy as np
def f(x): return x[0]**2+3*x[1]**2
x=np.array([2.,-1.])
analytic=np.array([2*x[0],6*x[1]])
eps=1e-6
numeric=np.array([(f(x+eps*np.eye(2)[i])-f(x-eps*np.eye(2)[i]))/(2*eps) for i in range(2)])
print("analytic:",analytic,"numeric:",numeric)
```

## Jacobian

For vector-valued $\mathbf{f}:\mathbb{R}^d\to\mathbb{R}^k$, the Jacobian is

$$
J_{ij}=\frac{\partial f_i}{\partial x_j}.
$$

Backpropagation repeatedly multiplies local Jacobians using the chain rule.

## Matrix derivative: least squares

For

$$
L(\mathbf{w})=\frac{1}{n}\|X\mathbf{w}-\mathbf{y}\|_2^2,
$$

the gradient is

$$
\nabla_{\mathbf{w}}L
=
\frac{2}{n}X^T(X\mathbf{w}-\mathbf{y}).
$$

Notice the shapes:

- $X$: $n\times d$
- residual: $n\times1$
- $X^T$ residual: $d\times1$, matching $\mathbf{w}$.

```python
rng=np.random.default_rng(0)
X=rng.normal(size=(8,3)); y=rng.normal(size=8); w=rng.normal(size=3)
grad=2/len(X)*X.T@(X@w-y)
print("gradient shape:",grad.shape)
```

## Chain rule

If

$$
L=L(z),\quad z=z(a),\quad a=a(x),
$$

then

$$
\frac{dL}{dx}
=
\frac{dL}{dz}
\frac{dz}{da}
\frac{da}{dx}.
$$

Neural-network backpropagation is this rule applied efficiently across a computational graph.

```python
import torch
x=torch.tensor(2.0,requires_grad=True)
a=x**2
z=torch.sin(a)
L=z**3
L.backward()
manual=3*(torch.sin(x.detach()**2)**2)*torch.cos(x.detach()**2)*(2*x.detach())
print("autograd:",x.grad.item(),"manual:",manual.item())
```
