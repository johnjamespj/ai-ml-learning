## Parameter estimation

A statistical model assumes data were generated from a distribution with unknown parameter $\theta$.

The likelihood is

$$
L(\theta)=p(\mathcal{D}\mid\theta).
$$

Maximum-likelihood estimation chooses

$$
\hat\theta_{MLE}=\arg\max_\theta p(\mathcal{D}\mid\theta).
$$

We usually maximize log-likelihood because products become sums.

## Gaussian MLE

If $x_i\sim\mathcal{N}(\mu,\sigma^2)$, then the MLE of $\mu$ is the sample mean.

```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(0)
x=rng.normal(2.5,1.2,80)
mu_grid=np.linspace(0,5,500)
sigma=1.2
logL=np.array([-0.5*np.sum((x-mu)**2/sigma**2) for mu in mu_grid])
mle=mu_grid[np.argmax(logL)]
print("sample mean",x.mean(),"grid MLE",mle)
plt.plot(mu_grid,logL); plt.axvline(x.mean(),ls="--"); plt.xlabel("mu"); plt.ylabel("log-likelihood"); plt.show()
```

## Sampling distribution and standard error

An estimator itself is random because the sample is random.

For the sample mean,

$$
SE(\bar X)=\frac{\sigma}{\sqrt{n}}.
$$

```python
for n in [10,30,100,1000]:
    means=[rng.normal(0,1,n).mean() for _ in range(5000)]
    print(n,np.std(means),"theory",1/np.sqrt(n))
```

## Bias and variance of an estimator

$$
\operatorname{MSE}(\hat\theta)
=
\operatorname{Bias}(\hat\theta)^2
+
\operatorname{Var}(\hat\theta).
$$

This decomposition echoes the model bias-variance tradeoff.
