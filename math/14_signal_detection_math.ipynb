## Binary hypothesis testing

A detector chooses between

$$
H_0: x=n
$$

and

$$
H_1: x=s+n.
$$

For known distributions, the Neyman-Pearson lemma says the most powerful test at a fixed false-alarm probability compares the likelihood ratio

$$
\Lambda(x)
=
\frac{p(x\mid H_1)}{p(x\mid H_0)}
$$

to a threshold $\eta$.

## Matched filter in white Gaussian noise

For known deterministic signal vector $s$ and

$$
n\sim\mathcal{N}(0,\sigma^2I),
$$

the log-likelihood ratio reduces, up to constants, to

$$
T(x)=s^Tx.
$$

So the matched filter is the optimal linear detector for this model.

```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(0)
N=64
s=np.sin(2*np.pi*.12*np.arange(N))
s=s/np.linalg.norm(s)
noise=rng.normal(size=(200000,N))
T0=noise@s
eta=np.quantile(T0,.99)
print("threshold for ~1% Pfa",eta,"empirical",np.mean(T0>eta))
for amp in [0,.5,1,1.5,2,3]:
    X=noise[:20000]+amp*s
    pd=np.mean((X@s)>eta)
    print("amplitude",amp,"Pd",pd)
```

## Unknown phase

If sinusoid phase is unknown, use quadrature projections

$$
I=x^Ts,\quad Q=x^Tc
$$

and energy

$$
T(x)=I^2+Q^2.
$$

This is related to a generalized likelihood-ratio test.

## ROC and operating point

A threshold does not create one universal detector quality. Sweeping it traces

$$
P_D(\eta)
\quad\text{vs}\quad
P_{FA}(\eta).
$$

Radar design often specifies acceptable $P_{FA}$ and then asks what $P_D$ is achievable at each SNR.

```python
from sklearn.metrics import roc_curve,auc
H0=rng.normal(size=(5000,N))
H1=rng.normal(size=(5000,N))+1.2*s
score=np.r_[H0@s,H1@s]; y=np.r_[np.zeros(len(H0)),np.ones(len(H1))]
fpr,tpr,thr=roc_curve(y,score)
print("AUC",auc(fpr,tpr))
plt.semilogx(np.clip(fpr,1e-5,1),tpr); plt.xlabel("P_FA"); plt.ylabel("P_D"); plt.grid(True); plt.show()
```
