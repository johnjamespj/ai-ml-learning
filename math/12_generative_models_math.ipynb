## VAE: latent-variable model

A VAE defines

$$
p_\theta(x,z)=p(z)p_\theta(x\mid z).
$$

The posterior $p_\theta(z\mid x)$ is generally intractable, so we introduce

$$
q_\phi(z\mid x).
$$

The evidence lower bound is

$$
\log p_\theta(x)
\ge
\mathbb{E}_{q_\phi(z\mid x)}[\log p_\theta(x\mid z)]
-
D_{KL}(q_\phi(z\mid x)\|p(z)).
$$

For diagonal Gaussian $q(z|x)=\mathcal{N}(\mu,\operatorname{diag}(\sigma^2))$ against standard normal:

$$
D_{KL}
=
-\frac12\sum_j
\left(
1+\log\sigma_j^2-\mu_j^2-\sigma_j^2
\right).
$$

```python
import torch
mu=torch.tensor([[1.,0.]])
logvar=torch.tensor([[0.,0.]])
kl=-.5*(1+logvar-mu.pow(2)-logvar.exp()).sum(1)
print("KL:",kl.item())
```

## GAN minimax objective

Original GAN objective:

$$
\min_G\max_D
\mathbb{E}_{x\sim p_{data}}[\log D(x)]
+
\mathbb{E}_{z\sim p(z)}[\log(1-D(G(z)))].
$$

At the optimal discriminator for fixed $G$,

$$
D^*(x)=\frac{p_{data}(x)}{p_{data}(x)+p_g(x)}.
$$

## Diffusion forward process

A common forward process is

$$
q(x_t\mid x_0)
=
\mathcal{N}
\left(
\sqrt{\bar\alpha_t}x_0,
(1-\bar\alpha_t)I
\right).
$$

So one can sample directly:

$$
x_t
=
\sqrt{\bar\alpha_t}x_0
+
\sqrt{1-\bar\alpha_t}\epsilon.
$$

```python
import numpy as np
x0=2.0; abar=.4
rng=np.random.default_rng(0)
xt=np.sqrt(abar)*x0+np.sqrt(1-abar)*rng.normal(size=100000)
print("empirical mean",xt.mean(),"theory",np.sqrt(abar)*x0)
print("empirical var",xt.var(),"theory",1-abar)
```
