## Singular value decomposition

Any matrix $X\in\mathbb{R}^{n\times d}$ can be written

$$
X=U\Sigma V^T.
$$

The singular values in $\Sigma$ quantify strength along orthogonal directions.

## PCA optimization view

For centered data, the first principal direction solves

$$
v_1
=
\arg\max_{\|v\|=1}
\operatorname{Var}(Xv).
$$

It is the eigenvector of the covariance matrix with largest eigenvalue, equivalently the first right singular vector of centered $X$.

```python
import numpy as np
rng=np.random.default_rng(0)
z=rng.normal(size=500)
X=np.c_[3*z+.3*rng.normal(size=500),z+.1*rng.normal(size=500)]
Xc=X-X.mean(0)
U,S,Vt=np.linalg.svd(Xc,full_matrices=False)
cov=Xc.T@Xc/(len(Xc)-1)
vals,vecs=np.linalg.eigh(cov)
print("SVD first direction:",Vt[0])
print("covariance top eigenvector:",vecs[:,-1])
print("sign-equivalent:",abs(Vt[0]@vecs[:,-1]))
```

## Kernel trick

Suppose a feature map $\phi(x)$ is high-dimensional. A kernel computes inner products

$$
k(x,x')=\phi(x)^T\phi(x')
$$

without explicitly constructing $\phi$.

For the RBF kernel,

$$
k(x,x')
=
\exp(-\gamma\|x-x'\|^2).
$$

```python
def rbf(x,y,gamma=1):
    return np.exp(-gamma*np.sum((x-y)**2))
print(rbf(np.array([0.,0.]),np.array([1.,1.])))
```
