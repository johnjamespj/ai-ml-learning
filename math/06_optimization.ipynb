## Optimization problem

Training often has the form

$$
\theta^*
=
\arg\min_\theta J(\theta).
$$

The objective may be convex, nonconvex, smooth, nonsmooth, deterministic, or stochastic.

## Gradient descent

$$
\theta_{t+1}=\theta_t-\eta\nabla J(\theta_t).
$$

The learning rate $\eta$ interacts with curvature. A step size safe in a flat direction may explode in a steep one.

```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
```

```python
import numpy as np, matplotlib.pyplot as plt
H=np.diag([1.,50.])
def f(x): return .5*x@H@x
def grad(x): return H@x
for eta in [.005,.03,.05]:
    x=np.array([8.,5.]); hist=[]
    for _ in range(100):
        hist.append(f(x)); x-=eta*grad(x)
    plt.plot(hist,label=f"eta={eta}")
plt.yscale("log"); plt.legend(); plt.title("Learning rate vs curvature"); plt.show()
```

## Hessian and curvature

The Hessian is

$$
H_{ij}=\frac{\partial^2J}{\partial\theta_i\partial\theta_j}.
$$

For a quadratic objective, Newton's update is

$$
\theta_{t+1}
=
\theta_t-H^{-1}\nabla J(\theta_t),
$$

which solves an exact quadratic in one step when $H$ is known and invertible.

```python
x=np.array([8.,5.])
x_new=x-np.linalg.solve(H,grad(x))
print("Newton result:",x_new,"objective:",f(x_new))
```

## Stochastic gradient

Instead of using all samples,

$$
\nabla J(\theta)
\approx
\frac{1}{|B|}\sum_{i\in B}\nabla\ell_i(\theta).
$$

Mini-batch noise can help exploration but also increases update variance.
