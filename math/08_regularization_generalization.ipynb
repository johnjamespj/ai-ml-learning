## Regularized empirical risk

A common objective is

$$
\hat R(\theta)+\lambda\Omega(\theta),
$$

where

$$
\hat R(\theta)
=
\frac{1}{n}\sum_{i=1}^n\ell(f_\theta(x_i),y_i).
$$

Regularization intentionally trades training fit for simpler or more stable solutions.

## L2 regularization

$$
\Omega(w)=\|w\|_2^2.
$$

For linear least squares:

$$
\hat w_{ridge}
=
(X^TX+\lambda I)^{-1}X^Ty.
$$

Adding $\lambda I$ improves conditioning.

```python
import numpy as np
rng=np.random.default_rng(0)
x=rng.normal(size=100)
X=np.c_[x,x+1e-3*rng.normal(size=100)]
y=3*x+rng.normal(0,.3,100)
for lam in [0,1e-4,.01,1]:
    A=X.T@X+lam*np.eye(2)
    w=np.linalg.solve(A,X.T@y)
    print("lambda",lam,"w",w,"norm",np.linalg.norm(w),"cond",np.linalg.cond(A))
```

## Bias-variance decomposition

For squared loss at a fixed $x$,

$$
\mathbb{E}[(y-\hat f(x))^2]
=
\sigma^2
+
\operatorname{Bias}[\hat f(x)]^2
+
\operatorname{Var}[\hat f(x)].
$$

Regularization can increase bias while lowering variance.

## Generalization gap

$$
\text{gap}=R_{test}-R_{train}.
$$

A small training loss is not evidence of a useful model unless the evaluation process approximates deployment.
