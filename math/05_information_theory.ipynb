## Self-information

An event with probability $p$ carries

$$
I(x)=-\log p(x)
$$

units of information. Rare events carry more surprise.

## Entropy

For a discrete distribution $p$,

$$
H(p)=-\sum_x p(x)\log p(x).
$$

Entropy measures average uncertainty.

```python
import numpy as np
def H(p):
    p=np.asarray(p,float); p=p[p>0]
    return -np.sum(p*np.log2(p))
for p in [[.5,.5],[.9,.1],[.99,.01]]:
    print(p,H(p),"bits")
```

## Cross-entropy

If data come from $p$ but the model predicts $q$,

$$
H(p,q)=-\sum_x p(x)\log q(x).
$$

For one-hot classification labels, minimizing cross-entropy means increasing probability on the correct class.

## KL divergence

$$
D_{KL}(p\|q)
=
\sum_x p(x)\log\frac{p(x)}{q(x)}
=
H(p,q)-H(p).
$$

It is nonnegative but **not symmetric**.

```python
def kl(p,q):
    p=np.asarray(p,float); q=np.asarray(q,float)
    return np.sum(p*np.log(p/q))
p=np.array([.8,.2]); q=np.array([.5,.5])
print("KL(p||q)",kl(p,q),"KL(q||p)",kl(q,p))
```

## Softmax and log-sum-exp

Softmax turns logits $z_k$ into probabilities:

$$
p_k=\frac{e^{z_k}}{\sum_j e^{z_j}}.
$$

Numerical stability comes from subtracting $\max_j z_j$ before exponentiation.

```python
z=np.array([1000.,999.,998.])
zs=z-z.max()
p=np.exp(zs)/np.exp(zs).sum()
print(p,p.sum())
```
