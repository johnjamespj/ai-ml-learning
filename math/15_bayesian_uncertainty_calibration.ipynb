## Bayes theorem for parameters

$$
p(\theta\mid\mathcal{D})
=
\frac{p(\mathcal{D}\mid\theta)p(\theta)}
{p(\mathcal{D})}.
$$

The posterior predictive distribution is

$$
p(y_*\mid x_*,\mathcal{D})
=
\int
p(y_*\mid x_*,\theta)
p(\theta\mid\mathcal{D})
\,d\theta.
$$

This integrates parameter uncertainty rather than using only one estimate.

## Beta-Bernoulli conjugacy

If

$$
\theta\sim\operatorname{Beta}(\alpha,\beta)
$$

and data contain $h$ successes and $t$ failures,

$$
\theta\mid\mathcal{D}
\sim
\operatorname{Beta}(\alpha+h,\beta+t).
$$

## Calibration

A probabilistic classifier is calibrated if

$$
P(Y=1\mid \hat p=p)=p.
$$

Brier score:

$$
\frac1n\sum_i(\hat p_i-y_i)^2.
$$

Calibration is different from discrimination: a model can rank cases well but report bad probabilities.

```python
from sklearn.metrics import brier_score_loss,roc_auc_score
rng=np.random.default_rng(0)
true_p=rng.uniform(0,1,10000)
y=rng.binomial(1,true_p)
good=true_p
overconf=np.clip((true_p-.5)*1.7+.5,0,1)
print("good AUC/Brier",roc_auc_score(y,good),brier_score_loss(y,good))
print("overconf AUC/Brier",roc_auc_score(y,overconf),brier_score_loss(y,overconf))
```
