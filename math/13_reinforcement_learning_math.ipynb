## Markov decision process

An MDP is $(\mathcal{S},\mathcal{A},P,R,\gamma)$.

The return is

$$
G_t
=
\sum_{k=0}^{\infty}\gamma^kR_{t+k+1}.
$$

## State value

Under policy $\pi$:

$$
V^\pi(s)
=
\mathbb{E}_\pi[G_t\mid S_t=s].
$$

Bellman expectation equation:

$$
V^\pi(s)
=
\sum_a\pi(a\mid s)
\sum_{s',r}
p(s',r\mid s,a)
\left[r+\gamma V^\pi(s')\right].
$$

## Optimal action value

$$
Q^*(s,a)
=
\mathbb{E}
\left[
R_{t+1}
+
\gamma\max_{a'}Q^*(S_{t+1},a')
\mid S_t=s,A_t=a
\right].
$$

Q-learning uses a stochastic approximation to this fixed point:

$$
Q(s,a)
\leftarrow
Q(s,a)
+
\alpha
\left[
r+\gamma\max_{a'}Q(s',a')-Q(s,a)
\right].
$$

```python
import numpy as np
Q=np.zeros((5,2))
s,a,r,ns=2,1,0.0,3
alpha=.2; gamma=.95
td_target=r+gamma*np.max(Q[ns])
td_error=td_target-Q[s,a]
Q[s,a]+=alpha*td_error
print("target",td_target,"TD error",td_error,"updated Q",Q[s,a])
```

## Contraction intuition

For $0\le\gamma<1$, the Bellman optimality operator is a contraction under the max norm. Repeated exact Bellman updates converge to a unique fixed point.

Function approximation breaks many of the tabular guarantees, which is why deep RL can be unstable.
