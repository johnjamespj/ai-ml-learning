## The mathematical objects

Most ML can be described using:

- scalar: $a\in\mathbb{R}$
- vector: $\mathbf{x}\in\mathbb{R}^d$
- matrix: $X\in\mathbb{R}^{n\times d}$
- tensor: an array with more than two axes
- function: $f:\mathbb{R}^d\to\mathbb{R}^k$
- parameter vector: $\theta$
- dataset: $\mathcal{D}=\{(\mathbf{x}_i,y_i)\}_{i=1}^n$

A habit that prevents many bugs: **annotate every important tensor with its shape before writing the operation.**

```python
import numpy as np
n,d,k=5,3,2
X=np.random.default_rng(0).normal(size=(n,d))
W=np.random.default_rng(1).normal(size=(d,k))
b=np.zeros(k)
Y=X@W+b
print("X:",X.shape,"W:",W.shape,"b:",b.shape,"Y:",Y.shape)
assert Y.shape==(n,k)
```

## Index notation

Matrix multiplication

$$
Y=XW
$$

means

$$
Y_{ij}=\sum_{r=1}^{d}X_{ir}W_{rj}.
$$

The repeated index $r$ is summed over. This is the same operation NumPy performs with `X @ W`.

```python
Y_manual=np.zeros((n,k))
for i in range(n):
    for j in range(k):
        for r in range(d):
            Y_manual[i,j]+=X[i,r]*W[r,j]
print(np.allclose(Y,Y_manual))
```

## Broadcasting

Broadcasting is convenient, but it can hide mathematical mistakes.

For $Y=XW+b$, the same bias vector $b\in\mathbb{R}^k$ is added to every sample row.

```python
print((X@W).shape,b.shape)
print("broadcast result:",(X@W+b).shape)
```
