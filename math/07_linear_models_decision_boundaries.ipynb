## Least squares geometry

Linear regression solves

$$
\min_w\|Xw-y\|_2^2.
$$

At the optimum,

$$
X^T(X\hat w-y)=0.
$$

So the residual is orthogonal to every column of $X$.

If $X^TX$ is invertible,

$$
\hat w=(X^TX)^{-1}X^Ty.
$$

Numerically, solving via QR/SVD is usually safer than explicitly forming the inverse.

```python
import numpy as np
rng=np.random.default_rng(0)
X=np.c_[np.ones(100),rng.normal(size=(100,2))]
wtrue=np.array([1.,2.,-1.])
y=X@wtrue+rng.normal(0,.2,100)
what=np.linalg.lstsq(X,y,rcond=None)[0]
r=y-X@what
print("w_hat",what)
print("X^T residual:",X.T@r)
```

## Logistic regression

Model probability:

$$
P(y=1\mid x)=\sigma(w^Tx+b)
$$

with

$$
\sigma(z)=\frac{1}{1+e^{-z}}.
$$

The decision boundary at threshold $0.5$ is

$$
w^Tx+b=0.
$$

A different probability threshold shifts the boundary.

```python
def logit(p): return np.log(p/(1-p))
for p in [.1,.5,.9]:
    print(p,"log-odds",logit(p))
```

## Margin

For a hyperplane $w^Tx+b=0$, signed distance is

$$
\frac{w^Tx+b}{\|w\|}.
$$

SVMs explicitly optimize a margin-related objective.
