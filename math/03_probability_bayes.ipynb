## Random variables

A random variable maps outcomes to numbers. We describe uncertainty with a probability distribution.

For a continuous variable $X$ with density $p(x)$:

$$
\mathbb{E}[X]=\int x p(x)\,dx
$$

and

$$
\operatorname{Var}(X)=\mathbb{E}[(X-\mathbb{E}[X])^2].
$$

## Conditional probability

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}.
$$

Bayes' rule reverses a conditional:

$$
P(A\mid B)
=
\frac{P(B\mid A)P(A)}{P(B)}.
$$

## Base-rate effect

Suppose:

- disease prevalence: $P(D)=0.01$
- sensitivity: $P(+\mid D)=0.95$
- false-positive rate: $P(+\mid \neg D)=0.05$

Then

$$
P(D\mid +)
=
\frac{0.95\cdot0.01}
{0.95\cdot0.01+0.05\cdot0.99}.
$$

```python
pD=.01; sens=.95; fpr=.05
posterior=sens*pD/(sens*pD+fpr*(1-pD))
print("P(disease | positive) =",posterior)
```

## Independence

$A$ and $B$ are independent if

$$
P(A,B)=P(A)P(B).
$$

Conditional independence is weaker and is central to graphical models and Naive Bayes.

```python
# Correlated Gaussian variables are not independent.
x=rng.normal(size=100000)
y=.8*x+.6*rng.normal(size=len(x))
print(np.corrcoef(x,y)[0,1])
```
