## Goal

## k-means
k-means alternates between:
1. assigning each sample to its nearest centroid;
2. replacing each centroid with the mean of its assigned samples.

It minimizes within-cluster squared distance.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(n_samples=500, centers=4, cluster_std=1.2, random_state=42)
model = KMeans(n_clusters=4, n_init="auto", random_state=42)
labels = model.fit_predict(X)
print(model.cluster_centers_)
```

```python
import numpy as np
from sklearn.datasets import make_moons, make_blobs
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
for name,(X,_) in {
    "blobs":make_blobs(n_samples=400,centers=3,random_state=1),
    "moons":make_moons(n_samples=400,noise=.06,random_state=1)
}.items():
    Xs=StandardScaler().fit_transform(X)
    km=KMeans(n_clusters=3,n_init="auto",random_state=0).fit(Xs)
    db=DBSCAN(eps=.25,min_samples=5).fit(Xs)
    print(name,"kmeans clusters",len(set(km.labels_)),"dbscan labels",sorted(set(db.labels_)))
```
