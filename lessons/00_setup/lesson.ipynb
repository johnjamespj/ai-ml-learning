## Concepts
A Python interpreter executes code. A virtual environment isolates project dependencies. A package is reusable Python software. Jupyter notebooks are excellent for experiments, while normal Python modules are preferable for reusable code.

## Setup
Clone the repository, enter it, then run:



````python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
````
