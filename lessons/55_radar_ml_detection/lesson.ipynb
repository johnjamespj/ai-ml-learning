```python
import numpy as np
rng=np.random.default_rng(0); N=128; f=.125; t=np.arange(N)
s=np.sin(2*np.pi*f*t); c=np.cos(2*np.pi*f*t)
def score(X):
    return ((X@s)**2+(X@c)**2)/N
noise=rng.normal(size=(10000,N))
threshold=np.quantile(score(noise),.99)
noise_test=rng.normal(size=(10000,N))
print("threshold for ~1% Pfa:",threshold,"fresh-test Pfa",np.mean(score(noise_test)>=threshold))
for snr_db in [-20,-15,-10,-5,0,5]:
    phase=rng.uniform(0,2*np.pi,3000)
    amp=np.sqrt(2*10**(snr_db/10))
    sig=amp*np.sin(2*np.pi*f*t[None,:]+phase[:,None])
    X=sig+rng.normal(size=sig.shape)
    print("SNR",snr_db,"Pd",np.mean(score(X)>=threshold))
```
