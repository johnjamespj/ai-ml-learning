- ROC
- precision-recall curve
- calibration

## Threshold sweep
A classifier score is not the final decision. Sweep thresholds and study operating points.

## Experiment
Build a synthetic detector where SNR spans -20 dB to +20 dB.

Report:
- Pd vs SNR
- false-alarm behavior
- confusion matrix
- calibration
- generalization to held-out frequencies

## Critical lesson
A single aggregate accuracy number hides the operating behavior engineers actually care about.


## Runnable activity
Run this experiment and change at least one data, threshold, deployment, or systems assumption.

```python
import numpy as np
rng=np.random.default_rng(0); N=128; f=.125; t=np.arange(N)
s=np.sin(2*np.pi*f*t); c=np.cos(2*np.pi*f*t)
def score(X):
    return ((X@s)**2+(X@c)**2)/N
noise=rng.normal(size=(10000,N))
threshold=np.quantile(score(noise),.99)
print("threshold for ~1% Pfa:",threshold,"empirical Pfa",np.mean(score(noise)>=threshold))
for snr_db in [-20,-15,-10,-5,0,5]:
    phase=rng.uniform(0,2*np.pi,3000)
    amp=np.sqrt(2*10**(snr_db/10))
    sig=amp*np.sin(2*np.pi*f*t[None,:]+phase[:,None])
    X=sig+rng.normal(size=sig.shape)
    print("SNR",snr_db,"Pd",np.mean(score(X)>=threshold))
```
