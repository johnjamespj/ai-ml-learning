```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
```

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance
from sklearn.model_selection import train_test_split
X,y=load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=0,stratify=y)
m=RandomForestClassifier(n_estimators=200,random_state=0).fit(Xtr,ytr)
r=permutation_importance(m,Xte,yte,n_repeats=8,random_state=0)
top=np.argsort(r.importances_mean)[::-1][:8]
print("top feature indices:",top)
print("importance means:",r.importances_mean[top])
x=Xte[0].copy(); base=m.predict_proba([x])[0,1]
for j in top[:3]:
    xp=x.copy(); xp[j]=np.median(Xtr[:,j])
    print("feature",j,"probability change",m.predict_proba([xp])[0,1]-base)
```
