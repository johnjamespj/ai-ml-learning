```python
import numpy as np, matplotlib.pyplot as plt
from scipy import signal
fs=2000
t=np.arange(0,2,1/fs)
x=signal.chirp(t, f0=100, f1=500, t1=2, method="linear")
f,tt,Z=signal.stft(x,fs=fs,nperseg=256,noverlap=192)
S=10*np.log10(np.abs(Z)**2+1e-12)
plt.pcolormesh(tt,f,S,shading="auto")
plt.ylim(0,1000); plt.xlabel("time (s)"); plt.ylabel("frequency (Hz)")
plt.title("Chirp spectrogram"); plt.colorbar(label="dB"); plt.show()
print("spectrogram shape",S.shape)
```
