# Lesson 48: Diffusion models

## Core idea
Diffusion models learn to reverse a gradual noising process.

Forward process:
clean sample -> progressively noisier samples

Learned reverse process:
noise -> progressively denoised sample

## Training intuition
A common formulation:
1. sample clean data x0;
2. sample a timestep t;
3. add known Gaussian noise;
4. ask a neural network to predict the noise;
5. optimize prediction error.

## Important concepts
- noise schedule
- timesteps
- denoising network
- score intuition
- latent diffusion
- conditional generation
- classifier-free guidance

## Why this matters
The same broad idea underlies many modern image-generation systems, but the training objective is much easier to understand on toy data.

```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(0)
x0=np.r_[rng.normal(-2,.35,500),rng.normal(2,.35,500)]
betas=np.linspace(1e-4,.08,60)
alpha_bar=np.cumprod(1-betas)
fig,axs=plt.subplots(1,4,figsize=(12,2.5))
for ax,t in zip(axs,[0,10,30,59]):
    eps=rng.normal(size=len(x0))
    xt=np.sqrt(alpha_bar[t])*x0+np.sqrt(1-alpha_bar[t])*eps
    ax.hist(xt,bins=40); ax.set_title(f"t={t}")
plt.tight_layout(); plt.show()
print("As t grows, information about the clean bimodal distribution is progressively destroyed.")
```
