```python
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0,x)

def softmax(x):
    shifted = x - np.max(x, axis=1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=1, keepdims=True)
```



Subtracting the maximum in softmax improves numerical stability.

```python
import numpy as np, matplotlib.pyplot as plt
x=np.linspace(-6,6,400)
sigmoid=lambda z:1/(1+np.exp(-z))
relu=lambda z:np.maximum(0,z)
plt.plot(x,sigmoid(x),label="sigmoid")
plt.plot(x,np.tanh(x),label="tanh")
plt.plot(x,relu(x),label="ReLU")
plt.legend(); plt.grid(True); plt.show()
big=np.array([[1000.,999.,998.]])
shifted=big-big.max(axis=1,keepdims=True)
p=np.exp(shifted)/np.exp(shifted).sum(axis=1,keepdims=True)
print("stable softmax:",p,p.sum())
```
