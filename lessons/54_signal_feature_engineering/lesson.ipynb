```python
import numpy as np
from scipy.signal import welch
rng=np.random.default_rng(0); fs=1000; t=np.arange(fs)/fs
signals={
 "tone":np.sin(2*np.pi*120*t)+.2*rng.normal(size=len(t)),
 "chirp":np.sin(2*np.pi*(60*t+180*t*t))+ .2*rng.normal(size=len(t)),
 "noise":rng.normal(size=len(t))
}
for name,x in signals.items():
    f,P=welch(x,fs=fs,nperseg=256)
    centroid=(f*P).sum()/P.sum()
    rms=np.sqrt(np.mean(x*x))
    peak=f[np.argmax(P)]
    entropy=-np.sum((P/P.sum())*np.log(P/P.sum()+1e-12))
    print(name,{"rms":rms,"peak_hz":peak,"centroid_hz":centroid,"spectral_entropy":entropy})
```
