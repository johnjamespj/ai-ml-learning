```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.calibration import calibration_curve
from sklearn.metrics import brier_score_loss
X,y=make_classification(n_samples=4000,n_features=15,class_sep=.8,flip_y=.08,random_state=0)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.4,random_state=1)
m=LogisticRegression(max_iter=1000).fit(Xtr,ytr)
p=m.predict_proba(Xte)[:,1]
pt,pp=calibration_curve(yte,p,n_bins=8)
print("Brier",brier_score_loss(yte,p))
for predicted,observed in zip(pp,pt):
    print(f"predicted {predicted:.2f} -> observed {observed:.2f}")
```
