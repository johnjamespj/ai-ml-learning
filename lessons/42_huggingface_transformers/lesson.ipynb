# Lesson 42: Hugging Face Transformers

Once you understand tokenization, embeddings, attention and language modeling, use higher-level libraries productively.

## Tokenizer and model


```python
try:
    from transformers import AutoTokenizer, AutoModelForSequenceClassification
    
    name = "distilbert/distilbert-base-uncased"
    tokenizer = AutoTokenizer.from_pretrained(name)
    model = AutoModelForSequenceClassification.from_pretrained(
        name,
        num_labels=2,
    )
    
    batch = tokenizer(
        ["signal detected", "noise only"],
        padding=True,
        truncation=True,
        return_tensors="pt",
    )
    
    logits = model(**batch).logits
    print(logits.shape)
except Exception as e:
    print("Pretrained-model example needs transformers + model access:", e)
```



## Learn the abstractions

## Runnable activity
This is a reduced-scale experiment for the core mechanism. Run it first, then extend it.

```python
try:
    from transformers import BertConfig, BertForSequenceClassification
    import torch
    cfg=BertConfig(vocab_size=100,hidden_size=32,num_hidden_layers=2,num_attention_heads=4,intermediate_size=64,num_labels=2)
    model=BertForSequenceClassification(cfg)
    ids=torch.randint(0,100,(4,12))
    mask=torch.ones_like(ids)
    out=model(input_ids=ids,attention_mask=mask)
    print("offline random-weight BERT logits:",out.logits.shape)
    print("parameters:",sum(p.numel() for p in model.parameters()))
except Exception as e:
    print("Install transformers to run this cell:",e)
# Once online, replace the random config with AutoTokenizer/AutoModel.from_pretrained(...)
```
