```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
X,y=make_moons(n_samples=600,noise=.2,random_state=42)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=42,stratify=y)
for kernel in ["linear","rbf"]:
    for C in [.1,1,10]:
        m=make_pipeline(StandardScaler(),SVC(kernel=kernel,C=C,gamma="scale"))
        m.fit(Xtr,ytr)
        print(kernel,"C",C,"test",m.score(Xte,yte))
```
