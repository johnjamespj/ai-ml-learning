# Lesson 45: Retrieval-Augmented Generation

RAG combines retrieval with generation.

## Basic pipeline
1. ingest documents
2. clean/parse
3. chunk
4. embed
5. index
6. embed query
7. retrieve candidates
8. optionally rerank
9. construct context
10. generate answer
11. retain citations/evidence

## Chunking
Chunk size and overlap influence retrieval behavior. Too small can lose context; too large can dilute matching and consume context window.

## Failure modes
- relevant chunk never indexed
- embedding mismatch
- poor query formulation
- wrong chunk retrieved
- correct chunk retrieved but ignored
- answer unsupported by retrieved evidence
- stale corpus
- context overflow

## Evaluation
Separate:
1. retrieval quality;
2. groundedness;
3. task answer quality;
4. latency/cost.

## Runnable activity
This is a reduced-scale experiment for the core mechanism. Run it first, then extend it.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np
docs=[
 "gradient descent updates parameters using the gradient of a loss",
 "attention uses queries keys and values",
 "PCA projects data onto directions of high variance",
 "RAG retrieves evidence before generation",
]
v=TfidfVectorizer().fit(docs)
M=v.transform(docs)
def retrieve(q,k=2):
    s=(M@v.transform([q]).T).toarray().ravel()
    idx=np.argsort(s)[::-1][:k]
    return [(int(i),docs[i],float(s[i])) for i in idx]
print(retrieve("How does retrieval augmented generation get evidence?"))
# Treat the retrieved text as context; retrieval quality can be evaluated before adding any generator.
```
