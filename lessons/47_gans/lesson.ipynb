# Lesson 47: Generative adversarial networks

A GAN has two competing models:

- Generator: produces synthetic examples
- Discriminator: tries to distinguish real from generated examples

The generator improves by learning to fool the discriminator.

## Concept
The discriminator sees:
- real x -> should output real
- G(z) -> should output fake

The generator learns from discriminator feedback.

## Main difficulties
- unstable training
- mode collapse
- sensitivity to architecture and hyperparameters
- difficult evaluation

## Important
GANs are powerful historically and conceptually, but many modern image-generation workflows rely heavily on diffusion models instead.


## Runnable activity
This is a reduced-scale experiment for the core mechanism. Run it first, then extend it.

```python
import torch
from torch import nn
torch.manual_seed(0)
G=nn.Sequential(nn.Linear(2,32),nn.ReLU(),nn.Linear(32,2))
D=nn.Sequential(nn.Linear(2,32),nn.ReLU(),nn.Linear(32,1))
og=torch.optim.Adam(G.parameters(),lr=.002); od=torch.optim.Adam(D.parameters(),lr=.002)
bce=nn.BCEWithLogitsLoss()
for step in range(400):
    real=torch.randn(128,2)*.3+torch.tensor([2.,0.])
    z=torch.randn(128,2); fake=G(z).detach()
    od.zero_grad(); ld=bce(D(real),torch.ones(128,1))+bce(D(fake),torch.zeros(128,1)); ld.backward(); od.step()
    z=torch.randn(128,2); og.zero_grad(); gen=G(z); lg=bce(D(gen),torch.ones(128,1)); lg.backward(); og.step()
with torch.no_grad():
    samp=G(torch.randn(2000,2))
print("generated mean",samp.mean(0).numpy(),"target approx [2,0]")
```
