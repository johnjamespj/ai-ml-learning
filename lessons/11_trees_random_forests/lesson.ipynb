# Lesson 11: Decision trees and random forests

A decision tree recursively splits feature space to make child groups purer.

## Classification tree


```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
X, y = make_classification(n_samples=500, n_features=6, n_informative=4, random_state=SEED)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.25, stratify=y, random_state=SEED)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(max_depth=4, random_state=42)
tree.fit(X_train, y_train)
```



Learn entropy, Gini impurity, information gain, depth, leaves, and pruning.

## Why trees overfit
A sufficiently deep tree can memorize tiny regions of training space.

## Random forests
A random forest trains many randomized trees and aggregates them. Bagging reduces instability/variance.



```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(
    n_estimators=300,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1,
)
forest.fit(X_train, y_train)
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
X,y=make_classification(n_samples=1200,n_features=12,n_informative=5,random_state=2)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=1,stratify=y)
for depth in [2,4,8,None]:
    t=DecisionTreeClassifier(max_depth=depth,random_state=0).fit(Xtr,ytr)
    print("tree depth",depth,"train",t.score(Xtr,ytr),"test",t.score(Xte,yte))
f=RandomForestClassifier(n_estimators=200,random_state=0,n_jobs=-1).fit(Xtr,ytr)
print("forest test",f.score(Xte,yte))
```
