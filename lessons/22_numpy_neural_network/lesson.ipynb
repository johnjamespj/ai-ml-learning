## Training loop skeleton


```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
from sklearn.datasets import make_moons
from labs.lab04_numpy_mlp.solution import init_params, forward, backward, cross_entropy
# This is the worked example. Implement your own version in the student lab.
X, y = make_moons(n_samples=200, noise=.2, random_state=SEED)
params = init_params(2, 16, 2, seed=SEED)
epochs = 300; learning_rate = .05
```

```python
for epoch in range(epochs):
    probs, cache = forward(X, params)
    loss = cross_entropy(probs, y)
    grads = backward(X, y, params, cache)

    for name in params:
        params[name] -= learning_rate * grads[name]
```

## Runnable activity
Execute, modify at least one parameter, and explain the observed change.

```python
import numpy as np
from sklearn.datasets import make_moons
X,y=make_moons(n_samples=300,noise=.2,random_state=1)
rng=np.random.default_rng(0)
W1=rng.normal(0,.5,(2,16)); b1=np.zeros(16); W2=rng.normal(0,.5,(16,2)); b2=np.zeros(2)
lr=.05
for epoch in range(1000):
    Z1=X@W1+b1; A1=np.maximum(0,Z1); logits=A1@W2+b2
    exp=np.exp(logits-logits.max(1,keepdims=True)); p=exp/exp.sum(1,keepdims=True)
    d=p.copy(); d[np.arange(len(y)),y]-=1; d/=len(y)
    dW2=A1.T@d; db2=d.sum(0); dA=d@W2.T; dZ=dA*(Z1>0); dW1=X.T@dZ; db1=dZ.sum(0)
    W1-=lr*dW1; b1-=lr*db1; W2-=lr*dW2; b2-=lr*db2
pred=p.argmax(1)
print("training accuracy:",(pred==y).mean())
```
