## Runnable activity
Execute, modify at least one parameter, and explain the observed change.

```python
import numpy as np
from sklearn.datasets import make_moons
X,y=make_moons(n_samples=300,noise=.2,random_state=1)
rng=np.random.default_rng(0)
W1=rng.normal(0,.5,(2,16)); b1=np.zeros(16); W2=rng.normal(0,.5,(16,2)); b2=np.zeros(2)
lr=.05
for epoch in range(1000):
    Z1=X@W1+b1; A1=np.maximum(0,Z1); logits=A1@W2+b2
    exp=np.exp(logits-logits.max(1,keepdims=True)); p=exp/exp.sum(1,keepdims=True)
    d=p.copy(); d[np.arange(len(y)),y]-=1; d/=len(y)
    dW2=A1.T@d; db2=d.sum(0); dA=d@W2.T; dZ=dA*(Z1>0); dW1=X.T@dZ; db1=dZ.sum(0)
    W1-=lr*dW1; b1-=lr*db1; W2-=lr*dW2; b2-=lr*db2
pred=p.argmax(1)
print("training accuracy:",(pred==y).mean())
```
