## Sliding windows
A simple sequence model can begin by converting a long series into fixed windows.



```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
```

```python
import numpy as np

def make_windows(x, window):
    X, y = [], []
    for i in range(len(x)-window):
        X.append(x[i:i+window])
        y.append(x[i+window])
    return np.array(X), np.array(y)
```

```python
import numpy as np
def make_windows(x,window):
    X=[]; y=[]
    for i in range(len(x)-window):
        X.append(x[i:i+window]); y.append(x[i+window])
    return np.asarray(X),np.asarray(y)
t=np.linspace(0,20,500)
s=np.sin(t)+.1*np.random.default_rng(0).normal(size=len(t))
X,y=make_windows(s,20)
print("window matrix",X.shape,"targets",y.shape)
baseline=X[:,-1]
print("persistence RMSE",np.sqrt(np.mean((baseline-y)**2)))
```
