# Lesson 18: Anomaly detection

Anomaly detection asks whether an observation is unusual relative to expected behavior.

## Why it is tricky
Anomalies are often rare, labels may be scarce, and "unusual" is not identical to "bad."

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.metrics import precision_score, recall_score
rng=np.random.default_rng(2)
normal=rng.normal(0,1,(1000,2))
anoms=rng.normal(5,.5,(30,2))
X=np.vstack([normal,anoms]); y=np.r_[np.zeros(len(normal)),np.ones(len(anoms))]
for contamination in [.01,.03,.08]:
    m=IsolationForest(contamination=contamination,random_state=0).fit(X)
    pred=(m.predict(X)==-1).astype(int)
    print(contamination,"precision",precision_score(y,pred),"recall",recall_score(y,pred))
```
