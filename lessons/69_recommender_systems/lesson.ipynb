# Lesson 69: Recommender systems

Recommendation is prediction over user-item interactions.

## Main approaches
- popularity baseline
- content-based filtering
- collaborative filtering
- matrix factorization
- neural recommenders

## Matrix factorization
Represent user u and item i with latent vectors:

score(u,i) = p_u dot q_i

Learn the vectors from observed interactions.

## Implicit vs explicit feedback
Ratings are explicit. Clicks, views and purchases are implicit and often ambiguous.

## Evaluation
Random splits can be misleading. Prefer time-aware evaluation when simulating future recommendations.

Metrics may include:
- Recall@k
- Precision@k
- MAP
- nDCG

## Runnable activity
Execute this reduced-scale experiment, then change one assumption and compare the result.

```python
import numpy as np
rng=np.random.default_rng(0)
n_users,n_items,k=60,80,5
trueU=rng.normal(size=(n_users,k)); trueV=rng.normal(size=(n_items,k))
R=trueU@trueV.T+.1*rng.normal(size=(n_users,n_items))
mask=rng.random(R.shape)<.25
U=rng.normal(scale=.1,size=(n_users,k)); V=rng.normal(scale=.1,size=(n_items,k))
lr=.02; reg=.01
rows,cols=np.where(mask)
for _ in range(50):
    for u,i in zip(rows,cols):
        e=R[u,i]-U[u]@V[i]
        u_old=U[u].copy()
        U[u]+=lr*(e*V[i]-reg*U[u]); V[i]+=lr*(e*u_old-reg*V[i])
rmse=np.sqrt(np.mean((R[mask]-(U@V.T)[mask])**2))
print("observed-entry RMSE",rmse)
print("top recommendations for user 0",np.argsort((U@V.T)[0])[::-1][:5])
```
