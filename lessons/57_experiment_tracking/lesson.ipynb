```python
from pathlib import Path
import os, sys
ROOT = Path(os.environ.get('COURSE_ROOT', next((str(p) for p in [Path.cwd(), *Path.cwd().parents] if (p / 'coursekit').is_dir()), '.'))).resolve()
if not (ROOT / 'coursekit').is_dir():
    raise RuntimeError('Open this notebook in the cloned ai-ml-learning repository.')
sys.path.insert(0, str(ROOT))
from coursekit.runtime import configure
SEED = 0
configure(SEED)
```

```python
import json, time, hashlib
from pathlib import Path
runs=Path("runs_demo"); runs.mkdir(exist_ok=True)
configs=[{"lr":.1,"seed":0},{"lr":.03,"seed":0},{"lr":.01,"seed":1}]
for cfg in configs:
    run_id=hashlib.sha1(json.dumps(cfg,sort_keys=True).encode()).hexdigest()[:8]
    metric=1/(1+abs(cfg["lr"]-.03)*10)+cfg["seed"]*.001
    record={"run_id":run_id,"config":cfg,"metric":metric,"timestamp":time.time()}
    (runs/f"{run_id}.json").write_text(json.dumps(record,indent=2))
print("saved",len(list(runs.glob("*.json"))),"structured run records")
display([json.loads(p.read_text()) for p in runs.glob("*.json")])
```
