## Runnable activity
Execute this reduced-scale experiment, then change one assumption and compare the result.

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score
rng=np.random.default_rng(0); n=4000
group=rng.integers(0,2,n)
y=rng.integers(0,2,n)
# Group 1 gets a noisier score to create unequal performance
score=y+rng.normal(0,np.where(group==0,.5,1.1))
pred=(score>.5).astype(int)
print("overall precision/recall",precision_score(y,pred),recall_score(y,pred))
for g in [0,1]:
    m=group==g
    print("group",g,"n",m.sum(),"precision",precision_score(y[m],pred[m]),"recall",recall_score(y[m],pred[m]))
```
