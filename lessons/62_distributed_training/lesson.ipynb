# Lesson 62: Distributed training

## Why distribute
Large models/datasets can exceed one device or one machine's practical throughput.

## Data parallelism
Replicate the model on multiple workers, split batches and synchronize gradients.

## Concepts
- process
- rank
- world size
- collective operations
- all-reduce
- DistributedDataParallel
- distributed sampler

## Beyond data parallelism
Learn the purpose of:
- model parallelism
- tensor parallelism
- pipeline parallelism
- sharding/FSDP concepts

## Important
Distributed training adds communication and coordination overhead. Two GPUs do not automatically mean 2x speed.

```python
import torch
print("torch.distributed available:",torch.distributed.is_available())
world_size=4
local_grads=[torch.tensor([float(rank+1),2.0]) for rank in range(world_size)]
simulated_allreduce=sum(local_grads)/world_size
print("local gradients:",local_grads)
print("average gradient after simulated all-reduce:",simulated_allreduce)
print("DDP performs this synchronization across worker processes while each worker handles different data.")
```
