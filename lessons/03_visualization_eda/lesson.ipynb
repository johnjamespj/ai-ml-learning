# Lesson 03: Visualization and exploratory data analysis

## Goal
Learn to inspect data before choosing a model.



```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
x = np.linspace(0, 10, 200)
y = 2*x + 1 + rng.normal(0, 2, len(x))

plt.scatter(x, y, alpha=0.6)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Synthetic observations")
plt.show()
```



Explore distributions, relationships, outliers, class balance and suspicious artifacts.

```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(42)
a=rng.normal([0,0],[1,1],(300,2))
b=rng.normal([2,1],[1.2,.8],(300,2))
X=np.vstack([a,b]); y=np.r_[np.zeros(len(a)),np.ones(len(b))]
plt.scatter(X[:,0],X[:,1],c=y,alpha=.45)
plt.title("Two overlapping classes")
plt.xlabel("feature 1"); plt.ylabel("feature 2"); plt.show()
print("class counts:",np.bincount(y.astype(int)))
print("correlation:\n",np.corrcoef(X.T))
```
