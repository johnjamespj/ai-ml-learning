## Generate data
Change noise and inspect how recovery changes.

```python
rng=np.random.default_rng(42)
X=rng.uniform(-5,5,(300,1))
y=3*X[:,0]-4+rng.normal(0,.5,300)
plt.scatter(X[:,0],y,s=10);
```

## Train
Before running: predict whether the first or last loss should be larger.

```python
w,b,history=lab.fit(X,y,lr=.01,steps=2000)
print(w,b,history[0],history[-1])
plt.figure(); plt.plot(history); plt.yscale('log'); plt.title('Training loss');
```
