## Central claim
Randomly dropping units during training can regularize neural networks and reduce co-adaptation/overfitting.

```python
import numpy as np, torch, matplotlib.pyplot as plt
from torch import nn
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
d=load_digits(); X=torch.tensor((d.data/16).astype("float32")); y=torch.tensor(d.target.astype("int64"))
tr,te=train_test_split(np.arange(len(y)),train_size=300,test_size=800,random_state=4,stratify=y.numpy())
Xtr,Xte,ytr,yte=X[tr],X[te],y[tr],y[te]
```

## Overparameterized network

```python
class Net(nn.Module):
    def __init__(self,p):
        super().__init__()
        self.net=nn.Sequential(nn.Linear(64,256),nn.ReLU(),nn.Dropout(p),
                               nn.Linear(256,256),nn.ReLU(),nn.Dropout(p),nn.Linear(256,10))
    def forward(self,x): return self.net(x)
def train(p):
    torch.manual_seed(0); m=Net(p); opt=torch.optim.Adam(m.parameters(),lr=.005); ce=nn.CrossEntropyLoss(); hist=[]
    for _ in range(120):
        m.train(); opt.zero_grad(); loss=ce(m(Xtr),ytr); loss.backward(); opt.step()
        m.eval()
        with torch.no_grad():
            train_acc=(m(Xtr).argmax(1)==ytr).float().mean().item()
            test_acc=(m(Xte).argmax(1)==yte).float().mean().item()
        hist.append([train_acc,test_acc])
    return np.array(hist)
a=train(0); b=train(.5)
print("no dropout train/test",a[-1]); print("dropout train/test",b[-1])
plt.plot(a[:,1],label="no dropout test"); plt.plot(b[:,1],label="dropout test"); plt.legend(); plt.show()
```
