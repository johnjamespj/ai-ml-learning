## Toy task family: unknown line y = ax + b
Each task provides demonstrations in context. No parameters are updated across queries.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
rng=np.random.default_rng(0)
def task():
    a=rng.uniform(-3,3); b=rng.uniform(-2,2)
    return a,b
def solve_from_context(xs,ys,xq):
    # TODO: replace polyfit with your own two-point derivation.
    if len(xs)<2: return np.mean(ys) if len(ys) else 0.0
    ah,bh=np.polyfit(xs,ys,1)
    return ah*xq+bh

rows=[]
for shots in [0,1,2,4]:
    errors=[]
    for _ in range(500):
        a,b=task(); xs=rng.uniform(-2,2,shots); ys=a*xs+b; xq=rng.uniform(-2,2); true=a*xq+b
        pred=solve_from_context(xs,ys,xq)
        errors.append((pred-true)**2)
    rows.append((shots,np.mean(errors)))
df=pd.DataFrame(rows,columns=["shots","MSE"])
display(df)
plt.plot(df.shots,df.MSE,marker="o"); plt.yscale("log"); plt.xlabel("examples in context"); plt.ylabel("MSE"); plt.show()
```

## What this does and does not show
It demonstrates why additional demonstrations can specify a task **without updating model weights**. It does not show that a language model learned to infer tasks from text.
