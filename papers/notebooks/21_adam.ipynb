```python
import numpy as np, matplotlib.pyplot as plt
def f(x): return x[0]**2+100*x[1]**2
def grad(x): return np.array([2*x[0],200*x[1]])
start=np.array([8.,6.])
```

## Implement SGD, momentum, and Adam

```python
def optimize(kind,steps=200,lr=.03):
    x=start.copy(); v=np.zeros(2); m=np.zeros(2); s=np.zeros(2); path=[x.copy()]; losses=[f(x)]
    for t in range(1,steps+1):
        g=grad(x)
        if kind=="sgd":
            x-=lr*g
        elif kind=="momentum":
            v=.9*v+g; x-=lr*v
        elif kind=="adam":
            m=.9*m+.1*g; s=.999*s+.001*g*g
            mh=m/(1-.9**t); sh=s/(1-.999**t)
            # TODO: rewrite this update from the Adam equations.
            x-=lr*mh/(np.sqrt(sh)+1e-8)
        path.append(x.copy()); losses.append(f(x))
    return np.array(path),np.array(losses)
results={}
for kind,lr in [("sgd",.004),("momentum",.004),("adam",.08)]:
    p,l=optimize(kind,lr=lr); results[kind]=(p,l); print(kind,"final",l[-1])
```

## Figure-inspired optimizer trajectories

```python
plt.figure(figsize=(6,5))
for kind,(p,l) in results.items():
    plt.plot(p[:,0],p[:,1],marker=".",markersize=2,label=kind)
plt.xlabel("x1"); plt.ylabel("x2"); plt.legend(); plt.title("Optimization paths"); plt.show()
for kind,(p,l) in results.items(): plt.plot(l,label=kind)
plt.yscale("log"); plt.legend(); plt.title("Objective value"); plt.show()
```
